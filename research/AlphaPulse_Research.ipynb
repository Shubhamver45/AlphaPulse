# AlphaPulse: Investment Risk & Volatility Research
This notebook contains the initial exploratory data analysis and mathematical prototyping for the AlphaPulse production pipeline.

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

tickers = ['AAPL', 'MSFT', 'NVDA', 'JNJ', 'JPM', 'AMZN']
data = yf.download(tickers, period='1y')['Close']
data.plot(figsize=(10, 6), title='Asset Price Discovery')
```

## 1. Log Returns and Volatility

```python
log_returns = np.log(data / data.shift(1)).dropna()
volatility = log_returns.std() * np.sqrt(252)
print("Annualized Volatility:")
print(volatility)
```

## 2. Monte Carlo Simulation Prototype

```python
S0 = 1000000
mu = log_returns.mean().mean()
sigma = log_returns.std().mean()
T = 252
n_sims = 1000

dt = 1/T
price_paths = np.zeros((T, n_sims))
price_paths[0] = S0

for t in range(1, T):
    z = np.random.standard_normal(n_sims)
    price_paths[t] = price_paths[t-1] * np.exp((mu - 0.5 * sigma**2) * dt + sigma * np.sqrt(dt) * z)

plt.plot(price_paths[:, :10])
plt.title('Monte Carlo Simulation (Sample Paths)')
```
